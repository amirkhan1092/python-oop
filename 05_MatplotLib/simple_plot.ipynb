```python
# simple custom 2D plot

import matplotlib.pyplot as plt

subject_marks = [90, 50, 70, 60, 80]

# Create a figure and axis
plt.plot(range(1, 6), subject_marks)

# Add title and labels
plt.title("Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

# Add grid
plt.grid()
# Add x and y ticks
plt.xticks(range(1, 6), ["Math", "English", "Science", "History", "Geography"])

# Add a legend
plt.legend(["Marks"], loc="upper right")
# Add a text on the plot each subject
for i, mark in enumerate(subject_marks):
    plt.text(i+1.05, mark, str(mark), fontsize=10, ha='left', va='top', color='red')



plt.show()
```

```python
'''
Matplotlib is a plotting library for the Python programming language and its numerical mathematics extension NumPy. 
It provides an object-oriented API for embedding plots into applications that use general-purpose GUI toolkits, such as Tkinter, wxPython, Qt, or GTK.'''

import matplotlib.pyplot as plt
import numpy as np

# Create some data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create a figure and axis
fig, ax = plt.subplots()
# Plot the data
ax.plot(x, y)
# Set the title and labels
ax.set_title('Sine Wave')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
# Show the plot
plt.show()
# Save the figure
fig.savefig('sine_wave.png')
```

```python
# matplotlib

import matplotlib.pyplot as plt

marks = [45, 60, 34, 78, 20, 50]
# subject = []

plt.plot(range(1, 7), marks, 'i')
plt.title("Khush Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid()
for i, j in enumerate(marks, 1):
    plt.text(i, j, str(j))
plt.xticks(range(1, 7), ['Math', 'Computer', 'Web', 'Physics', 'English', 'Design'])
plt.axhline(y=40 ,color='red', linewidth=2, label='Fail line (40)')
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [0, 1, 2, 3, 4]
y1 = [i for i in x]
y2 = [i**2 for i in x]

plt.plot(x, y1, label='y = x')
plt.plot(x, y2, label='y = x^2')
plt.title("Multiple Lines")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [200, 220, 250, 230, 240, 260]

plt.plot(months, sales, color='orange')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales')
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.show()
```
