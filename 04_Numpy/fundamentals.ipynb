```python
# matrix in Numpy 
M1 = [[2, 4, 5], [30, 10, 4], [0, 4, -3]]
M2 = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

Amat1 = np.array(M1)
Amat2 = np.array(M2)

if Amat1.shape == Amat2.shape:
    res = np.add(Amat1, Amat2)
    print('M1')
    print(Amat1)
    print('\nM2')
    print(Amat2)
    print('\nResult: Addition')
    print(res)
else:
    print('Invalid Dimension for the Addition')
```
