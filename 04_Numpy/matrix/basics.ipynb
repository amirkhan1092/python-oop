```python
# matrix 
class array:
    def __init__(self, lst):
        self.lst = lst
        
    def __str__(self):
        st = str(self.lst).replace(',', '')
        return st
    def __add__(self, other):
        adder = []
        if len(self.lst) == len(other.lst):
            for i in range(len(self.lst)):
                adder.append(self.lst[i]+other.lst[i])
            return array(adder)
        else:
            return "Length is not same"
    def reshape(self, m, n):
        M = []
        tmp = []
        for i in self.lst:
            tmp.append(i)
            if len(tmp) == n:
                M.append(tmp)
                tmp = []
        return array(M)

arr1 = array([2, 4, 6, 1, 6, 2])
arr2 = array([4, 7, 9, 0, 2, 0])
out = arr1.reshape(3, 2)
print(out)

```

```python
def isValidParenthesisWithWildcard(s):
    low = high = 0  # Tracks the possible range of open parentheses
    for char in s:
        if char == '(':
            low += 1
            high += 1
        elif char == ')':
            low -= 1
            high -= 1
        elif char == '*':
            low -= 1  # Treat '*' as ')'
            high += 1  # Treat '*' as '('

        # Adjust low since it can't be negative (too many ')')
        if low < 0:
            low = 0

        # If high < 0, too many closing brackets
        if high < 0:
            return "false"

    return "true" if low == 0 else "false"

# Sample test driver
if __name__ == "__main__":
    # n = int(input())
    s = input().strip()
    print(isValidParenthesisWithWildcard(s))
```

```python
from typing import List

class Order:
    def __init__(self, order_id: str, customer_name: str, amount: int):
        self.order_id = order_id
        self.customer_name = customer_name
        self.amount = amount

    @classmethod
    def is_premium_customer(cls, orders: List["Order"]) -> List[str]:
        premium = set()
        for order in orders:
            if order.amount > 10000:
                premium.add(order.customer_name)
        return sorted(premium)

if __name__ == "__main__":
    n = int(input())
    orders = []
    for _ in range(n):
        order_id, customer_name, amount = input().split()
        orders.append(Order(order_id, customer_name, int(amount)))

    result = Order.is_premium_customer(orders)
    if result:
        print(" ".join(result))
    else:
        print([])
```
